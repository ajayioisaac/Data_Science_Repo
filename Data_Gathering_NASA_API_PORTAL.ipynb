```python
import requests
import pandas as pd
from datetime import date

API_KEY = "9n9g7vYJaJoKLW1gul0fzlWuspl20DNwRXS8Wl27"

apod_url = f"https://api.nasa.gov/planetary/apod?api_key={API_KEY}"

# Make the request
response = requests.get(apod_url)
data = response.json()

# Display the title and image
print("Title:", data["title"])
print("Date:", data["date"])
print("Explanation:", data["explanation"])
print("Image URL:", data["url"])

from IPython.display import Image, display

display(Image(url=data["url"]))

today = date.today().isoformat()
neo_url = f"https://api.nasa.gov/neo/rest/v1/feed?start_date={today}&end_date={today}&api_key={API_KEY}"

response = requests.get(neo_url)
neo_data = response.json()

# Collect all asteroids for today
asteroids = []
for date_key in neo_data['near_earth_objects']:
    for asteroid in neo_data['near_earth_objects'][date_key]:
        asteroids.append({
            'Asteroid ID': asteroid['id'],
            'Asteroid name': asteroid['name'],
            'Minimal estimated diameter (km)': asteroid['estimated_diameter']['kilometers']['estimated_diameter_min'],
            'Absolute_magnitude': asteroid['absolute_magnitude_h'],
            'Relative_velocity(km/s)': float(asteroid['close_approach_data'][0]['relative_velocity']['kilometers_per_second']) if asteroid['close_approach_data'] else None
        })

# Create DataFrame
df = pd.DataFrame(asteroids)

# Display first few rows
df.head()
```
