```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd
import plotly.express as px


# Toy dataset
toy_dataset = [
    ['Skirt', 'Sneakers', 'Scarf', 'Pants', 'Hat'],
    ['Sunglasses', 'Skirt', 'Sneakers', 'Pants', 'Hat'],
    ['Dress', 'Sandals', 'Scarf', 'Pants', 'Heels'],
    ['Dress', 'Necklace', 'Earrings', 'Scarf', 'Hat', 'Heels', 'Hat'],
    ['Earrings', 'Skirt', 'Skirt', 'Scarf', 'Shirt', 'Pants']
]

# Encode the transactions
te = TransactionEncoder()
te_ary = te.fit(toy_dataset).transform(toy_dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

# Run Apriori
frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)

print("Frequent Itemsets:")
print(frequent_itemsets)

print("\nAssociation Rules:")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
# Load dataset
df = pd.read_csv("Market_Basket_Optimisation.csv")  # replace with actual file path

# Preview
print(df.head())

# Check for nulls and basic info
print(df.info())

# Optional visualization
item_counts = df.stack().value_counts().head(20)
px.bar(x=item_counts.index, y=item_counts.values, labels={'x': 'Items', 'y': 'Frequency'}, title='Top 20 Most Frequent Items').show()
```

```python
# Convert rows into lists
transactions = df.values.tolist()

# Remove NaNs (if any)
transactions = [[str(item) for item in row if pd.notna(item)] for row in transactions]

# Encode
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# Run Apriori
frequent_itemsets = apriori(df_encoded, min_support=0.05, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.1)

print("Top Rules:")
print(rules.sort_values(by='lift', ascending=False).head())
```
