```python
class BankAccount:

    def __init__ (self, name, balance = 0):
        self.name = name
        self.balance = balance
    
    def deposit(self, d_amount):
        self.balance += d_amount
        print (f"{self.name} deposited: {d_amount}. New balance: {self.balance}.")
    
    def withdraw(self, w_amount):
        if w_amount > self.balance:
            print("Insufficient Balance")
        else:
            self.balance -= w_amount
            print(f"{self.name} withdraws: {w_amount}. New balance: {self.balance}.")
    
    def get_balance(self):
        return f"Account Balance for {self.name} is {self.balance}"
    
    def display_info(self):
        print( f"{self.name}'s account balance is {self.balance}")

tolu = BankAccount("Tolu", 500)
tolu.deposit(200)
tolu.withdraw(800)
tolu.withdraw(300)
tolu.display_info()
```

```python
class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def introduce(self):
        print (f"Hello, my name is {self.name} and I am {self.age} years old.")

class SchoolMember(Person):

    def __init__(self, name,age, SchoolName):
        super().__init__(name, age)
        self.SchoolName = SchoolName

    def show_school(self):
        print(f"I am a member of {self.SchoolName} school")

class Student(SchoolMember):

    def __init__(self, name, age, SchoolName, grade):
        super().__init__(name,age, SchoolName)
        self.grade = grade
    
    def show_grade(self):
        print(f"I am in grade {self.grade}")
```

```python
# Data Encoding

import pandas as pd
#1. Using data mapping
cat_data = {"Animal": ["Dog", "Cat", "Dog", "Bird", "Cat", "Fish", "Rat", "Dog", "Cat", "Bird"]
            }
df =pd.DataFrame(cat_data)
animal_mapping = {"Dog": 0, "Cat": 1, "Bird": 2, "Fish": 3, "Rat": 4}
df['Animal'] = df['Animal'].map(animal_mapping)

#2. Using label Encoder Return ordinal datas
from sklearn.preprocessing import LabelEncoder 
df_2 = pd.DataFrame(cat_data)
label_encoder = LabelEncoder()
df_2["Animal"] = label_encoder.fit_transform(df_2["Animal"])
df["Animal"] = label_encoder.inverse_transform(df["Animal"]) # to reverse to categorical form


#3 One Hot Encoding Returns nominal datas
#a. use get dummies from pandas or one hot encoder from sklearn
d3 =pd.get_dummies(df, columns=["Animal"], dtype=int)

#b. Using OneHotEncoder from sklearn
from sklearn.preprocessing import OneHotEncoder 
encode = OneHotEncoder()
encoded_data = encode.fit_transform(df[["Animal"]]).toarray()
# Convert to DataFrame
encoded_df = pd.DataFrame(encoded_data, columns=encode.get_feature_names_out(["Animal"]))

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from datetime import datetime 

# set random seed for reproducibility
np.random.seed(42)

# Generate random data
dates = pd.date_range(start="2023-01-01", periods=500, freq='D')
categories = ['Electronics', 'Clothing', 'Home', 'Books', 'Toys']
regions = ['North', 'South', 'East', 'West']

sales_data =pd.DataFrame({
    'Date': np.random.choice(dates, 500),
    'Category': np.random.choice(categories, 500),
    'Region': np.random.choice(regions, 500),
    'Sales': np.random.randint(100, 5000, 500),
    'Profit': np.random.uniform(10, 500, 500),
    'Customer_Rating': np.random.uniform(1, 5, 500)
})

# Add some trends
sales_data.loc[sales_data['Category'] == 'Electronics', 'Sales'] *= 1.5
sales_data.loc[sales_data['Region'] == 'North', 'Profit'] *= 1.2
sales_data['Month'] = sales_data['Date'].dt.month_name()

monthly_sales = sales_data.groupby(['Month', 'Category']).agg({'Sales': 'sum'}).reset_index()
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
monthly_sales['Month'] = pd.Categorical(monthly_sales['Month'], categories=month_order, ordered=True)
monthly_sales = monthly_sales.sort_values('Month')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split    
```

```python
# Define the independent variables (House Size, Bedrooms, House Age)
house_features = np.array([
    [50, 2, 10],  
    [75, 3, 5],  
    [100, 3, 8],  
    [125, 4, 4],  
    [150, 4, 2]
])
# Define the dependent variable (House Price in $)
house_prices = np.array([150000, 200000, 250000, 300000, 350000])
```

```python
# Initialize the Linear Regression model
model = LinearRegression()
# Train the model using the dataset
model.fit(house_features, house_prices)
```

```python
# Get coefficients and intercept
m_values = model.coef_  # Slopes
b = model.intercept_  # Intercept
# Display the equation
print(f"Equation of the regression line: Y = {m_values[0]:.2f}X1 + {m_values[1]:.2f}X2 + {m_values[2]:.2f}X3 + {b:.2f}")
```

```python
# New house data for prediction (Size, Bedrooms, Age)
new_house = np.array([[120, 3, 6]])
# Predict the price
predicted_price = model.predict(new_house)
# Display prediction
print(f"Predicted House Price: ${predicted_price[0]:.2f}")
```

```python
# Plot House Size vs Price
plt.scatter(house_features[:, 0], house_prices, color='blue', label='Actual Prices')  # Actual prices
plt.scatter(new_house[:, 0], predicted_price, color='red', label='Predicted Price', marker='x', s=100)  # Predicted price
plt.xlabel("House Size (m²)")
plt.ylabel("House Price ($)")
plt.title("Effect of House Size on Price")
plt.legend()
plt.show()
```
