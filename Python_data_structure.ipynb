Question 1: Write a Python program that multiplies all the items in a list. 
Sample list = [2, 3, 6]
Expected Result = 36

```python
sample = [2, 3, 6]
total = 1
for i in sample:
    total *=i
print(f"The product of the items in {sample} is {total}") 
```

Question 2: Write a Python program to get a list, sorted in increasing order by the last element in each tuple, from a given list of non-empty tuples

```python
list_ = [(2,5), (1,2), (4,4), (2,3), (2,1)]
list_.sort(key=lambda i: i[-1])
print(f"Sorted list in increasing order by the last element in each tuple {list_}")
```

Question 3: Write a Python program that combines two dictionaries by adding values for common keys

```python
d1 = {
    'a': 100,
    'b': 200,
    'c': 300
}
d2 = {
    'a': 300,
    'b': 200,
    'd': 400
}
combine = {}
for i in d1:
    for j in d2:
        if i == j:
            combine.update({i: d1[i]+d2[j]})    
        if i not in combine:
            combine.update({i: d1[i]})
        if j not in combine:
            combine.update({j: d2[j]})
print(f"Printing the combination of two dictionaries by adding values for common keys\nList 1: {d1}      List 2: {d2} \nCombined List {combine}")
```

Question 4: With a given integral number n, write a program to generate a dictionary that contains (i, i*i) so that is an integral number between 1 and n (both included). Then the program should print the dictionary. 

```python
integral = int(input('Enter a end number: '))
dico = {}
for i in range(1, integral+1):
    dico[i] = i*i

print (dico)
```

Question 5: Write a Program to sort a tuple by its float element

```python
list1 = [('item1', '12.20'), ('item2', '15.10'), ('item3', '24.5')]
list1.sort(key=lambda x: float(x[1]), reverse=True)
print(list1)
```
