```python
# Doing data cleaning on the data received
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load and clean data
data = pd.read_csv("FLO PHD DATA.csv")
data.columns = data.columns.str.strip()  # Clean column names

# Select features and target by index
features = data.columns[15:24]
target = data.columns[13]

for feature in features:
    sns.scatterplot(data=data, x=feature, y=target)
    plt.title(f"{feature} vs {target}")
    plt.tight_layout()
    plt.show()
```
