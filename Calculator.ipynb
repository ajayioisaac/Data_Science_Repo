```python
import math

class Calculator:
    def __init__(self):
        # Dictionary to store operations
        self.operations = {
            '+': lambda x, y: x + y,
            '-': lambda x, y: x - y,
            '*': lambda x, y: x * y,
            '/': lambda x, y: x / y if y != 0 else float('inf')  # avoid ZeroDivisionError
        }

    def add_operation(self, symbol, function):
        self.operations[symbol] = function

    def calculate(self, num1, operation, num2=None):
        try:
            if not isinstance(num1, (int, float)) or (num2 is not None and not isinstance(num2, (int, float))):
                raise ValueError("Inputs must be numbers.")
            if operation not in self.operations:
                raise ValueError("Invalid operation.")

            if num2 is None:  # for single-operand operations like sqrt, log
                return self.operations[operation](num1)
            return self.operations[operation](num1, num2)
        except Exception as e:
            print(f"Error: {e}")
            raise

# Advanced operations
def power(x, y):
    return math.pow(x, y)

def square_root(x):
    if x < 0:
        raise ValueError("Cannot take square root of negative number.")
    return math.sqrt(x)

def logarithm(x):
    if x <= 0:
        raise ValueError("Logarithm only defined for positive numbers.")
    return math.log(x)

# Main program
if __name__ == "__main__":
    calc = Calculator()
    # Add advanced operations
    calc.add_operation('^', power)
    calc.add_operation('sqrt', square_root)
    calc.add_operation('log', logarithm)

    print("Welcome to the Calculator!")
    print("Available operations: +, -, *, /, ^, sqrt, log")

    while True:
        try:
            op = input("Enter operation (+, -, *, /, ^, sqrt, log or 'exit' to quit): ").strip()
            if op.lower() == 'exit':
                print("Goodbye!")
                break

            num1 = float(input("Enter the first number: "))
            num2 = None
            if op not in ['sqrt', 'log']:
                num2 = float(input("Enter the second number: "))

            result = calc.calculate(num1, op, num2)
            print(f"Result: {result}")

        except Exception:
            print("Something went wrong. Please try again.")
```
