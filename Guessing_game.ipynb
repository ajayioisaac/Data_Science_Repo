```python
""" Guess the Number is a single player game wher the player is prompted to gues a number between 1 and 100. The game
provides feedback to the player if their guess is too high or too low until they correctly guess the number."""

import random  # Importing random module

# Generate a random number between 1 and 100
guess = random.randint(1, 100)

print("...Welcome to the Guess the Number game!...")
print("\nI'm thinking of a number between 1 and 100. Can you guess what it is?")
# Loops until the user guesses correctly
while True:
    try:
        # Asking the user for input
        user_guess = int(input("Enter your guess: "))

        # Difference calculation
        difference = user_guess - guess

        if difference == 0:
            print("Congratulations! Your guess is correct.")
            break  # Exit the loop
        elif difference > 20:
            print("Your guess is too high. Guess again.")
        elif difference > 10:
            print("Your guess is still high. Guess again.")
        elif difference > 0:
            print("Your guess is high but close. Guess again.")
        elif difference < -20:
            print("Your guess is too low. Guess again.")
        elif difference < -10:
            print("Your guess is still low. Guess again.")
        else:  # difference < 0 and difference >= -10
            print("Your guess is low but close. Guess again.")

    except ValueError:
        print("Invalid input! Please enter a valid number.")

```
